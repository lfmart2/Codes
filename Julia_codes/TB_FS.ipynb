## Pre-processing
This section contains all the functions to load the Tight-Binding hopping elements calculated through AB-inition (VASP) and wannierization (Wannier90) from the file seedname_hr.dat. Here, we construct the hopping elements via HRData which contains the information of the number of WFs, constructed supercell, degeneracy on each unitcell, and the complex hopping elements $t$ and $t'$. We are projecting the energy states in a basis of d-orbitalas and sp-orbitals for the Nb and P atoms, respectively:
$$ \bra{\phi^{(\text{A}_i)}_{\alpha}} H_{\text{tot}} \ket{\phi^{(\text{B}_j)}_{\beta}} $$
where $\{\text{A}_i, \text{B}_j\} refer to the atoms Nb and P with label $i$ and $j$ of the supercell, and $\alpha$, $\beta$ are the atomic orbitals of each atom.
